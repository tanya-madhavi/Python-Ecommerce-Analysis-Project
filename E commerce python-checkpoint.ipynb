                   E-Commerce Data Analysis & Customer Behavior Insights

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
customers = pd.read_csv("CUSTOMERS.csv")
geo_location = pd.read_csv("GEO_LOCATION.csv")
order_items = pd.read_csv("ORDER_ITEMS.csv")
order_payments = pd.read_csv("ORDER_PAYMENTS.csv")
order_ratings = pd.read_csv("ORDER_REVIEW_RATINGS.csv")
orders = pd.read_csv("ORDERS.csv")
products = pd.read_csv("PRODUCTS.csv")
sellers = pd.read_csv("SELLERS.csv")
print("all files successfully loaded")
```

```python
order_items.columns
```

```python
orders = orders.drop_duplicates()
order_items = order_items.drop_duplicates()
customers = customers.drop_duplicates()
products = products.drop_duplicates()
sellers = sellers.drop_duplicates()
order_payments = order_payments.drop_duplicates()
order_ratings = order_ratings.drop_duplicates()
```

```python
total_revenue = order_items['price'].sum()
total_products = products['product_id'].nunique()
total_categories = products['product_category_name'].nunique()
total_sellers = sellers['seller_id'].nunique()
total_locations = customers['customer_state'].nunique()
total_channels = order_payments['payment_type'].nunique()
print("Total Revenue:",total_revenue)
print("Total Unique Products:",total_products)
print("Total Categories:",total_categories)
print("Total Unique Sellers:",total_sellers)
print("Total Unique Locations:",total_locations)
print("Total payment Channel:",total_channels)
```

```python
orders['order_purchase_timestamp'] = pd.to_datetime(orders['order_purchase_timestamp'])
orders['order_month'] = orders['order_purchase_timestamp'].dt.to_period('M')
customer_first_order = orders.groupby('customer_id')['order_month'].min().reset_index()
customer_first_order.columns = ['customer_id','first_purchase_month']
new_customer_by_month = customer_first_order.groupby('first_purchase_month').size().reset_index(name = 'new_customer_count')
```

```python
new_customer_by_month.head(5)
```

INSIGHTS
Early Growth Trajectory: Initial months (2016-09 to 2016-12) showed low customer acquisition (e.g., only 4 in Sept 2016 and 1 in Dec 2016), indicating early testing or platform launch phases.

Acquisition Surge: Customer acquisition spiked drastically in Jan 2017 (800 new users) and more than doubled in Feb 2017 (1,780 new users), signaling successful platform scaling and marketing push.

```python
customer_orders = pd.merge(orders,customer_first_order,on = 'customer_id')
customer_orders['total_months'] = (customer_orders['order_month'].astype(int)-customer_orders['first_purchase_month'].astype(int))
cohort_matrix = customer_orders.pivot_table(index = 'first_purchase_month',columns = 'total_months',values = 'customer_id',aggfunc = 'nunique')
```

```python
cohort_matrix.head(8)
```

```python
plt.figure(figsize=(12, 8))
sns.heatmap(cohort_matrix, annot=True, fmt='.0f', cmap='YlGnBu')
plt.title('Customer Retention Cohort Matrix')
plt.xlabel('Months Since First Purchase')
plt.ylabel('Cohort Join Month')
plt.show()
```

INSIGHTS
Methodology: Groups users into monthly cohorts based on their join date (first_purchase_month) and tracks their subsequent activity over time (total_months).

Business Purpose: This forms the core logic for Customer Retention & Churn Analysis to check how many unique users return in Month 1, Month 2, Month 3, etc

```python
order_revenues = pd.merge(order_items, orders, on='order_id')
order_revenues = pd.merge(order_revenues, customer_first_order, on='customer_id')
order_revenues['customer_type'] = 'Existing'
order_revenues.loc[order_revenues['order_month'] == order_revenues['first_purchase_month'], 'customer_type'] = 'New'
revenue_split = order_revenues.pivot_table(
    index='order_month',
    columns='customer_type',
    values='price',
    aggfunc='sum'
).fillna(0) 
```

```python
time_data = pd.merge(order_items, orders, on='order_id')
time_data['order_purchase_timestamp'] = pd.to_datetime(time_data['order_purchase_timestamp'])
time_data['order_month'] = time_data['order_purchase_timestamp'].dt.to_period('M')
time_data['day_of_week'] = time_data['order_purchase_timestamp'].dt.day_name()
time_data['hour_of_day'] = time_data['order_purchase_timestamp'].dt.hour
monthly_sales = time_data.groupby('order_month')['price'].sum().reset_index()
weekday_sales = time_data.groupby('day_of_week')['price'].sum().reset_index().sort_values(by='price', ascending=False)
hourly_sales = time_data.groupby('hour_of_day')['price'].sum().reset_index().sort_values(by='price', ascending=False)
print("--- TOP MONTHS BY SALES ---")
print(monthly_sales.head(5))
print("\n--- TOP DAYS BY SALES ---")
print(weekday_sales)
print("\n--- TOP HOURS BY SALES (Peak Hours) ---")
print(hourly_sales.head(5))

```

```python
import warnings
warnings.filterwarnings('ignore')
customer_revenue = pd.merge(order_items, orders, on='order_id')
customer_monetary = customer_revenue.groupby('customer_id')['price'].sum().reset_index(name='Total_Revenue')
labels = ['Bronze', 'Silver', 'Gold', 'Platinum']
customer_monetary['Customer_Group'] = pd.qcut(customer_monetary['Total_Revenue'], q=[0, 0.3, 0.7, 0.9, 1.0], labels=labels)
group_summary = customer_monetary.groupby('Customer_Group', observed=False).agg(
    customer_count=('customer_id', 'count'),
    total_group_revenue=('Total_Revenue', 'sum'),
    average_revenue_per_customer=('Total_Revenue', 'mean')
).reset_index()
print("--- CUSTOMER SEGMENTATION SUMMARY ---")
print(group_summary)
plt.figure(figsize=(10, 6))
sns.barplot(
    data=group_summary, 
    x='Customer_Group', 
    y='customer_count', 
    hue='Customer_Group', 
    palette='YlOrRd', 
    legend=False
)
plt.title('Distribution of Customers Across Revenue Groups', fontsize=14, fontweight='bold')
plt.xlabel('Customer Segments', fontsize=12)
plt.ylabel('Number of Customers', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

INSIGHTS
The Platinum group (top 10%) drives the maximum financial impact, generating $5.57M (over 40% of total revenue) despite comprising only 9,823 customers.
Platinum customers exhibit an average spend of $567.45, which is roughly 17x higher than Bronze tier customers ($31.98).

```python
seller_revenue = order_items.groupby('seller_id')['price'].sum().reset_index(name='Total_Revenue')
labels = ['Bronze', 'Silver', 'Gold', 'Platinum']
seller_revenue['Seller_Group'] = pd.qcut(seller_revenue['Total_Revenue'], q=[0, 0.3, 0.7, 0.9, 1.0], labels=labels)
group_summary_seller = seller_revenue.groupby('Seller_Group', observed=False).agg(
    seller_count=('seller_id', 'count'),
    total_group_revenue=('Total_Revenue', 'sum'),
    average_revenue_per_seller=('Total_Revenue', 'mean')
).reset_index()
print("--- SELLER SEGMENTATION SUMMARY ---")
print(group_summary_seller)
import warnings
warnings.filterwarnings('ignore')
plt.figure(figsize=(10, 6))
sns.barplot(
    data=group_summary_seller, 
    x='Seller_Group', 
    y='seller_count', 
    hue='Seller_Group', 
    palette='Purples_r', 
    legend=False
)
plt.title('Distribution of Sellers Across Revenue Groups', fontsize=14, fontweight='bold')
plt.xlabel('Seller Segments', fontsize=12)
plt.ylabel('Number of Sellers', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
from itertools import combinations
from collections import Counter
df_multi_prod = order_items[order_items.duplicated(subset=['order_id'], keep=False)]
order_product_lists = df_multi_prod.groupby('order_id')['product_id'].apply(list)
top_10_products = order_items['product_id'].value_counts().head(10).index.tolist()
pair_counter = Counter()
for products_in_order in order_product_lists:
 
    filtered_prods = [p for p in products_in_order if p in top_10_products]
  
    filtered_prods = sorted(list(set(filtered_prods)))
    
   
    if len(filtered_prods) >= 2:
        for comb in combinations(filtered_prods, 2):
            pair_counter[comb] += 1


top_10_combos = pair_counter.most_common(10)

import pandas as pd
cross_sell_df = pd.DataFrame(
    [(f"Product A: {item[0][0][:10]}...", f"Product B: {item[0][1][:10]}...", item[1]) for item in top_10_combos],
    columns=['Product_A', 'Product_B', 'Times_Bought_Together']
)

print("--- TOP 10 CROSS-SELLING COMBINATIONS (Bought Together) ---")
print(cross_sell_df)
```

```python
payment_method_counts = order_payments.groupby('payment_type').agg(
    customer_count=('order_id', 'nunique'),
    total_payment_value=('payment_value', 'sum')
).reset_index().sort_values(by='customer_count', ascending=False)

print("--- PAYMENT BEHAVIOR SUMMARY ---")
print(payment_method_counts)

import warnings
warnings.filterwarnings('ignore')

fig, ax = plt.subplots(1, 2, figsize=(16, 6))
sns.barplot(
    data=payment_method_counts, 
    x='payment_type', 
    y='customer_count', 
    hue='payment_type', 
    palette='Set2', 
    legend=False,
    ax=ax[0]
)
ax[0].set_title('Number of Customers per Payment Method', fontsize=12, fontweight='bold')
ax[0].set_xlabel('Payment Method')
ax[0].set_ylabel('Unique Customers (Orders)')
ax[0].grid(axis='y', linestyle='--', alpha=0.5)

ax[1].pie(
    payment_method_counts['total_payment_value'], 
    labels=payment_method_counts['payment_type'], 
    autopct='%1.1f%%', 
    startangle=140, 
    colors=sns.color_palette('Set2')
)
ax[1].set_title('Total Value Share (%) per Payment Method', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
category_review_df = pd.merge(order_items, order_ratings, on='order_id')
category_review_df = pd.merge(category_review_df, products, on='product_id')

cat_summary = category_review_df.groupby('product_category_name').agg(
    avg_rating=('review_score', 'mean'),
    review_count=('review_score', 'count')
).reset_index()
cat_filtered = cat_summary[cat_summary['review_count'] >= 10]

top_10_max_categories = cat_filtered.sort_values(by='avg_rating', ascending=False).head(10)
top_10_min_categories = cat_filtered.sort_values(by='avg_rating', ascending=True).head(10)
print("==========================================")
print("   TOP 10 MAXIMUM RATED CATEGORIES")
print("==========================================")
display(top_10_max_categories)

print("\n==========================================")
print("   TOP 10 MINIMUM RATED CATEGORIES")
print("==========================================")
display(top_10_min_categories)
```

```python
product_review_df = pd.merge(order_items, order_ratings, on='order_id')
prod_summary = product_review_df.groupby('product_id').agg(
    avg_rating=('review_score', 'mean'),
    review_count=('review_score', 'count')
).reset_index()

prod_filtered = prod_summary[prod_summary['review_count'] >= 10]

top_10_max_products = prod_filtered.sort_values(by='avg_rating', ascending=False).head(10)
top_10_min_products = prod_filtered.sort_values(by='avg_rating', ascending=True).head(10)

print("==========================================")
print("   TOP 10 MAXIMUM RATED PRODUCTS")
print("==========================================")
display(top_10_max_products)

print("\n==========================================")
print("   TOP 10 MINIMUM RATED PRODUCTS")
print("==========================================")
display(top_10_min_products)
```

indicating exceptional quality control and high satisfaction for specific SKUs.

```python
import warnings
warnings.filterwarnings('ignore')
master_review = pd.merge(order_items, order_ratings, on='order_id')
master_review = pd.merge(master_review, products, on='product_id')
master_review = pd.merge(master_review, orders, on='order_id')
master_review = pd.merge(master_review, customers, on='customer_id')
location_rating = master_review.groupby('customer_state')['review_score'].mean().reset_index()
location_rating = location_rating.sort_values(by='review_score', ascending=False)
seller_rating = master_review.groupby('seller_id').agg(
    avg_rating=('review_score', 'mean'),
    total_reviews=('review_score', 'count')
).reset_index()
seller_rating = seller_rating[seller_rating['total_reviews'] >= 20].sort_values(by='avg_rating', ascending=False)
master_review['order_purchase_timestamp'] = pd.to_datetime(master_review['order_purchase_timestamp'])
master_review['order_month_str'] = master_review['order_purchase_timestamp'].dt.to_period('M').astype(str)
month_rating = master_review.groupby('order_month_str')['review_score'].mean().reset_index()
print("==========================================")
print("   AVERAGE RATING BY LOCATION (TOP 10 STATES)")
print("==========================================")
display(location_rating.head(10))

print("\n==========================================")
print("   TOP RATED SELLERS (Min 20 Reviews)")
print("==========================================")
display(seller_rating.head(10))

print("\n==========================================")
print("   AVERAGE RATING BY MONTH")
print("==========================================")
display(month_rating.head(10))

plt.figure(figsize=(10, 4))
sns.lineplot(data=month_rating, x='order_month_str', y='review_score', marker='o', color='purple', linewidth=2)
plt.title('Average Customer Rating Trend by Month', fontsize=12, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Average Review Score')
plt.xticks(rotation=45)
plt.grid(axis='both', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
